# Demo: Single-Fidelity UQ for Farmer's Problem with 1000 possible scenarios

This demo shows how to run the StandardMRP algorithm (single-fidelity) on the 
Advanced farmer's problem with a given first-stage candidate solution.

The output is a point-estimate and confidence interval for an 
upper bound on the optimality gap.

```python
from sparow.conf_intervals.options import UQOptions
from sparow.conf_intervals.standard_mrp import StandardMRP
from sparow.conf_intervals.evaluate_true_optimality_gap import TrueOptimalityGapEvaluator

# This is the function you have to define for your problem instance
from sparow_examples.farmers.MRPfarmers import get_sp_model_for_uq 
```

```python
# STEP 1 - Get the model object for the farmer's problem
model = get_sp_model_for_uq(
    model_name="Advanced",
    use_integer=False,
    seed=12345,
    with_replacement=True,
)

# STEP 2 - Define the first-stage candidate solution (xhat) to evaluate
xhat = {
    "DevotedAcreage[WHEAT]": 100.0,
    "DevotedAcreage[CORN]": 80.0,
    "DevotedAcreage[SUGAR_BEETS]": 320.0,
}

# STEP 3 - Define what options you want to use for the MRP algorithm
options = UQOptions(
    n=25, # number of scenarios to sample per replication
    m=30, # number of replications to perform
    alpha=0.05, # significance level for the confidence interval
    seed=12345, # random seed for reproducibility of algorithm run
    with_replacement=True, # whether to sample scenarios with replacement or not from finite population
    solver_name="gurobi_direct", 
    verbose=False,
)

# STEP 4 - Run the MRP algorithm with the model, options, and xhat solution
mrp_algorithm = StandardMRP(model=model, options=options)
results = mrp_algorithm.run(xhat=xhat)

print("\nMRP results:")
print(f"MRP Point estimate: {results['point_estimate']}")
print(f"CI: [{results['ci_lower']}, {results['ci_upper']}]")
print(f"Sample Variance: {results['sample_variance']}")
```

```python
# Since this is a relatively small toy problem, you can solve the extensive form over all scenarios
# Compute and display the true finite-population optimality gap
true_gap_evaluator = TrueOptimalityGapEvaluator(
    model=model,
    solver_name="gurobi_direct",
    solver_options=None,
)

true_gap_results = true_gap_evaluator.compute_true_gap(xhat=xhat)

print("\nTrue finite-population quantities:")
print(f"True optimal value: {true_gap_results['true_optimal_value']}")
print(f"xhat true value: {true_gap_results['xhat_true_value']}")
print(f"True optimality gap: {true_gap_results['true_gap']}")
```
